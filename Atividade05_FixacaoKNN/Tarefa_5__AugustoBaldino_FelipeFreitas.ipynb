```python
import pandas as pd
from sklearn import model_selection, neighbors
from math import floor
```

```python
integrantes = ['Felipe Freitas', 'Augusto Baldino']
```

```python
dividas_full = pd.read_excel('sample_data/datasetDividas.xlsx')
dataLinesCount = 21
# dividas_full.head(dataLinesCount)
```

```python
# Identifique os atributos relevantes para a atividade.
dividas = dividas_full.drop(columns=['cliente'])
# dividas.head(dataLinesCount)
```

```python
# Crie uma versão 2 do dataset com os valores dos atributos relevantes normalizados.
# Eles devem ficar entre 0 e 1.
# Uma estratégia simples é identificar o maior valor de cada coluna relevante
#   e dividir todos os valores da referida coluna pelo seu maior valor.
maiorRenda = dividas['renda'].max()
maiorDivida = dividas['divida'].max()
dividas['renda'] = dividas['renda'].div(maiorRenda)
dividas['divida'] = dividas['divida'].div(maiorDivida)
dividas.head(dataLinesCount)
```

```python
train_set, test_set = model_selection.train_test_split(dividas, train_size=0.7)
```

```python
classifier = neighbors.KNeighborsClassifier(n_neighbors=5)

train_set_x = train_set.drop(columns=['classe'])
train_set_y = train_set['classe']

classifier.fit(train_set_x, train_set_y)
```

```python
test_set_x = test_set.drop(columns=['classe'])
test_set_y = test_set['classe'].values

predicao = classifier.predict(test_set_x)

print("Predição | Fato")
for i in range(0, len(test_set)):
  print(predicao[i], " | ", test_set_y[i])
```
